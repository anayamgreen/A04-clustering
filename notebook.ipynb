````python
# Load in data
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

def createData(noise,N=50):
    np.random.seed(100) # Set the seed for replicability
    # Generate (x1,x2,g) triples:
    X1 = np.array([np.random.normal(1,noise,N),np.random.normal(1,noise,N)])
    X2 = np.array([np.random.normal(3,noise,N),np.random.normal(2,noise,N)])
    X3 = np.array([np.random.normal(5,noise,N),np.random.normal(3,noise,N)])
    # Concatenate into one data frame
    gdf1 = pd.DataFrame({'x1':X1[0,:],'x2':X1[1,:],'group':'a'})
    gdf2 = pd.DataFrame({'x1':X2[0,:],'x2':X2[1,:],'group':'b'})
    gdf3 = pd.DataFrame({'x1':X3[0,:],'x2':X3[1,:],'group':'c'})
    df = pd.concat([gdf1,gdf2,gdf3],axis=0)
    return df

df0_125 = createData(0.125)
df0_25 = createData(0.25)
df0_5 = createData(0.5)
df1_0 = createData(1.0)
df2_0 = createData(2.0)


# Scatter plots:
# Noise = 0.125
fig, ax = plt.subplots()
sns.scatterplot(data=df0_125, x="x1", y="x2", hue="group", ax=ax)
ax.set_title("Noise= 0.125")
plt.show()
#0.25
fig, ax = plt.subplots()
sns.scatterplot(data=df0_25, x="x1", y="x2", hue="group", ax=ax)
ax.set_title("Noise= 0.25")
plt.show()

#0.5
fig, ax = plt.subplots()
sns.scatterplot(data=df0_5, x="x1", y="x2", hue="group", ax=ax)
ax.set_title("Noise= 0.5")
plt.show()

#1.0
fig, ax = plt.subplots()
sns.scatterplot(data=df1_0, x="x1", y="x2", hue="group", ax=ax)
ax.set_title("Noise= 1.0")
plt.show()

#2.0
fig, ax = plt.subplots()
sns.scatterplot(data=df2_0, x="x1", y="x2", hue="group", ax=ax)
ax.set_title("Noise= 2.0")
plt.show()


#Scree plot .125


dataset = {
    "0.125": df0_125,
    "0.25": df0_25,
    "0.5": df0_5,
    "1.0": df1_0,
    "2.0": df2_0

}

# print(dataset)


# example set
# X = df0_125[["x1","x2"]]
# ks = np.arange(1, 11)
# SSE = [
#     KMeans(n_clusters=k, max_iter=300, n_init=10, random_state=0)
#     .fit(X).inertia_
#     for k in ks
# ]
# sns.lineplot(x=ks, y=SSE).set(title="Scree  0.125", xlabel="k", ylabel="SSE")


for name, df in dataset.items():

    X = df[["x1", "x2"]]

    SSE = [
    KMeans(n_clusters=k, max_iter=300, n_init=10, random_state=0)
    .fit(X).inertia_
    for k in ks
    ]
    sns.lineplot(x=ks, y=SSE, label=name)

plt.title("Scree Plots")
plt.xlabel("k")
plt.ylabel("SSE")
plt.legend()
plt.show()


````

4. Explain the intuition of the elbow, using this numerical simulation as an example.


The intuition of the elbow is that as we increase k, the SSE will decrease because there are more clusters to fit the data. In this simulation, there are three main groups, so going from 1 to 3 clusters causes a large decrease in SSE because the model is better separating into the original groups. After k = 3, adding more clusters still decreases SSE, but by a reuced amount because the three main groups have already been placed into their groups. This is what creates the elbow effect of the plot.
